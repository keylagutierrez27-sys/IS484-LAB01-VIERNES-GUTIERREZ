## Ficha del Problema: Desplazamiento del Repartidor
### Parte A: : Completar la cha del problema
* **Estado inicial:** `"Almacen"`
* **Estado objetivo:** `"ZonaNorte"
* **Acciones posibles:** 
  * Desplazarse entre los puntos interconectados de la ciudad: *Almacén, Centro, Terminal, Mercado y Zona Norte* según las rutas disponibles en el mapa vial.
* **Costo de cada acción:** 
  * 1 paso o unidad de distancia por cada tramo o arista recorrida entre un punto de la ciudad y el siguiente.
* **Solución (Secuencia de acciones esperada):** 
  * `Almacen` $\rightarrow$ `Centro` $\rightarrow$ `ZonaNorte` (o rutas alternativas equivalentes según el algoritmo de búsqueda).

### Parte B: Representación del mapa de la ciudad como un grafo

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

#### Explicacion de Creación del mapa

Este bloque crea un **mapa de conexiones** entre diferentes lugares.

- `mapa = {}`: Crea un diccionario para almacenar los lugares y sus conexiones.
- `"Almacen"`: Indica que está conectado con **Centro** y **Terminal**.
- `"Centro"`: Indica que está conectado con **Almacen, ZonaNorte y Mercado**.
- `"Terminal"`: Indica que está conectado con **Almacen y Mercado**.
- `"Mercado"`: Indica que está conectado con **Centro, Terminal y ZonaNorte**.
- `"ZonaNorte"`: Indica que está conectado con **Centro y Mercado**.
- `print(mapa)`: Muestra el mapa completo en pantalla.

**Resultado:** Se obtiene un mapa donde cada lugar tiene registrados los lugares a los que está conectado.


### Implementacion de DFS

```python
def dfs(grafo, inicio, objetivo):
    """Busca un camino entre 'inicio' y 'objetivo' usando DFS."""
    pila = [[inicio]]
    visitados = set()
    while pila:
        camino = pila.pop()  # sacar el ULTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]
        if nodo_actual == objetivo:
            return camino
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: misma logica que en BFS, pero agregando a la pila
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)
    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

#### Explicacion de Búsqueda DFS

Este bloque de código utiliza el algoritmo **DFS (búsqueda en profundidad)** para encontrar un camino entre dos lugares del mapa.

- `def dfs(...)`: Crea la función que realizará la búsqueda DFS.
- `pila`: Guarda los caminos que se van encontrando durante la búsqueda.
- `visitados`: Registra los lugares que ya fueron revisados.
- `while pila`: Continúa la búsqueda mientras existan caminos disponibles.
- `camino.pop()`: Toma el último camino de la pila, siguiendo la lógica **LIFO**.
- `nodo_actual`: Obtiene el lugar en el que se encuentra actualmente.
- `if nodo_actual == objetivo`: Comprueba si se llegó al destino.
- `for vecino`: Revisa los lugares conectados al lugar actual.
- `nuevo_camino`: Crea un camino agregando el siguiente lugar.
- `pila.append()`: Agrega el nuevo camino a la pila.
- `resultado_dfs`: Ejecuta la búsqueda desde `Almacen` hasta `ZonaNorte`.
- `print()`: Muestra el camino encontrado.


### vectores y matrices con NumPy

```python
import numpy as np

# Parte A: Representación de cuatro observaciones (temperatura y humedad)
obs1 = np.array([22.5, 60])  # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

dataset = np.array([obs1, obs2, obs3, obs4])
print("Dataset completo:\n", dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)

# Parte B: Acceso a filas y columnas específicas
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

#### Explicacion de Representación de datos con NumPy

Este bloque utiliza **NumPy** para crear y organizar datos de temperatura y humedad en una matriz.

- `import numpy as np`: Importa NumPy para trabajar con arreglos y matrices.
- `obs1`, `obs2`, `obs3`, `obs4`: Representan cuatro observaciones de temperatura y humedad.
- `dataset`: Une las cuatro observaciones en una sola matriz.
- `print("Dataset completo...")`: Muestra todos los datos almacenados.
- `dataset.shape`: Muestra el tamaño de la matriz, indicando filas y columnas.
- `dataset[0]`: Muestra la primera observación.
- `dataset[:, 0]`: Obtiene todas las temperaturas, que corresponden a la primera columna.

**Resultado:** Se crea una matriz con **4 observaciones y 2 características: temperatura y humedad**.
