### Importación de Bibliotecas

Este bloque de código realiza la carga inicial de los paquetes fundamentales para el curso:

* **`import numpy as np`**: Carga NumPy para el manejo de arreglos multidimensionales y operaciones matemáticas vectorizadas.
* **`import pandas as pd`**: Carga Pandas para la manipulación, limpieza y estructuración de datos en tablas (DataFrames).
* **`import matplotlib`**: Carga la librería base Matplotlib para la visualización de datos y generación de gráficos.
* **`import sklearn`**: Carga Scikit-learn, la biblioteca principal para la implementación de algoritmos de Machine Learning e Inteligencia Artificial.
* **`import scipy`**: Carga SciPy para realizar cálculos científicos, funciones estadísticas avanzadas y optimización

```python
import numpy as np
import pandas as pd
import matplotlib
import sklearn
import scipy
```

### Impresión de Versiones

Este bloque de código valida la instalación de las bibliotecas en el entorno activo[cite: 1]:

* **`np.__version__`**: Accede al atributo de NumPy que almacena la versión instalada en la memoria[cite: 1].
* **`pd.__version__`**: Consulta el número de versión de Pandas.
* **`matplotlib.__version__`**: Consulta la versión de Matplotlib instalada.
* **`sklearn.__version__`**: Consulta la versión de Scikit-learn.
* **`scipy.__version__`**: Consulta la versión de SciPy.
* **`print(...)`**: Muestra en la salida de la consola el nombre de cada librería junto a su versión para verificar que todo esté correctamente instalado y operativo.


```python
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
print("Scikit-learn:", sklearn.__version__)
print("SciPy:", scipy.__version__)
```
